# 04 - Overrides

Overrides are business rules that run after classification.

The classifier suggests. The system decides.

```python
def apply_overrides(text, intent, dialogue_act=None):
    text = text.lower()

    if dialogue_act == "acknowledgement":
        return "neutral"

    if "course list" in text or "ki ki course" in text:
        if intent in {"course_list", "career_guidance", "neutral", "unknown"}:
            return "course_list"

    if "confused" in text or "kon course" in text:
        if intent in {"neutral", "unknown"}:
            return "career_guidance"

    return intent

apply_overrides("ami confused, kon course nibo?", "unknown")
```

```python
apply_overrides("thanks", "course_details", dialogue_act="acknowledgement")
```

```python
apply_overrides("ami confused, kon course nibo?", "neutral")
```
