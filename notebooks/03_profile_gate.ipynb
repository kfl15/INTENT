# 03 - Profile Gate

A business profile says which intents are trusted for that business.

Even if the model returns a label confidently, the backend rejects it when it is outside the active profile.

Cause, 
LLM should choose the intent from the provided intent_list. 
But there is a slight possibility that the LLM could create a bad intent outside the list. 
So for safety, Backend still verifies.
If outside list -> marks as "unknown"

```python
allowed_intents = {
    "course_list",
    "course_pricing_payment",
    "career_guidance",
    "neutral",
    "irrelevant",
    "unknown",
}

def allowed_intent_gate(predicted_intent):
    if predicted_intent not in allowed_intents:
        return {
            "intent": "unknown",
            "confidence": "low",
            "reason": f"{predicted_intent} is outside the trusted profile.",
        }
    return {"intent": predicted_intent, "confidence": "high", "reason": "Allowed."}

allowed_intent_gate("bank_loan_application")
```
