# 02 - Local Fallback

The AI path is the primary classifier, but production systems need graceful degradation.

If the API key is missing or the provider fails, local keyword rules can still return a useful intent.

```python
# A local fallback rule is easy to inspect and test.
keyword_rules = {
    "course_pricing_payment": ["price", "fee", "cost", "payment", "koto"],
    "recording_access": ["record", "recording", "lms"],
    "course_schedule_duration": ["schedule", "batch", "duration", "class kobe"],
}

def local_classify(text):
    text = text.lower()
    for intent, keywords in keyword_rules.items():
        if any(keyword in text for keyword in keywords):
            return {"intent": intent, "confidence": "high", "source": "local"}
    return {"intent": "unknown", "confidence": "low", "source": "local"}

local_classify("ML course er price koto?")
```
