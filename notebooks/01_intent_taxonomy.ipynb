# 01 - Intent Taxonomy

In intent detection, we first decide the fixed list of labels the system is allowed to use.

Important idea from class: the model predicts an intent only. The backend derives category.

```python
# These are the 19 intents from the class deck, grouped by meaning.
course_info = [
    "course_list",
    "course_details",
    "course_pricing_payment",
    "course_projects",
    "course_schedule_duration",
    "career_certificate_outcomes",
]

fit_eligibility = [
    "prerequisite_fit",
    "beginner_friendliness",
    "background_fit",
    "career_guidance",
]

support_action = [
    "demo_trial_counseling",
    "enrollment_admission",
    "student_support_redirect",
    "instructor_info",
    "recording_access",
    "missed_class_recovery",
]

system_states = ["neutral", "irrelevant", "unknown"]

intents = course_info + fit_eligibility + support_action + system_states
len(intents), intents
```

```python
# Category is not another AI prediction.
# It is a simple dictionary lookup. This is safer, faster, and testable.
intent_to_category = {intent: "relevant" for intent in course_info + fit_eligibility + support_action}
intent_to_category.update({"neutral": "neutral", "irrelevant": "irrelevant", "unknown": "unknown"})

def category_for_intent(intent):
    return intent_to_category.get(intent, "unknown")

category_for_intent("course_pricing_payment"), category_for_intent("neutral")
```
